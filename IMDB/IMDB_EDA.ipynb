```python
from datetime import datetime


def fix_country(d):
    if pd.isna(d):
        return d
    if d.year > datetime.now().year:
        d = d.replace(year = d.year - 100)
    return d
```

```python
def safe_parse(s):
    try:
        return parse(s,fuzzy=True)
    except(ValueError,OverflowError):
        return pd.NaT
```
