```python
import re
def clean_score(s):

    if pd.isna(s):
        return s
    digits = re.sub(r'\D' , '', str(s))
    interger_part = digits[0]
    float_part = digits[1:] if len(digits) > 1 else '0'
    return float(f"{interger_part}.{float_part}")
```

```python
df['Score'] = df['Score'].apply(clean_score)
```
